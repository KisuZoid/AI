- Arbitrary Arguments: Allow function to accept multiple values.
    - *args : collects extra positional arguments as a tuple
    - **kwargs : collect extra keyword arguments as a dictionary

```python
# *args example : args is just a name, * followed by any name. *k, *arg, *argv etc
def fun(*args): 
    return sum(args)

print(fun(5, 10, 15))

# **kwargs : kwargs is just a name, ** followed by any name. **k, **kwarg, **arg etc
def fun(**kwargs):
    for k, val in kwargs.items():
        print(k, val)

fun(a=1, b=2, c=3)
```

```python
def studentInfo(*args, **kwargs):
    print("Subject:", args)
    print("Details:", kwargs)

# we can loop through the arguments as well --> prints all value separately
def loop(*args):
    for element in args:
        print(element)

studentInfo("Maths", "Science", "English", Name = "Kislay", Age = "22", City = "Delhi")

loop("Kislay", "Anand")
```

```python
# Function within Function
def f1():
    S = "I love F1 cars"
    def f2():
        print(S)
    f2()

f1()
```

### Pass by Reference and Pass by Value

** Python uses pass-by-object-reference, where functions receive references to objects instead of actual copies.
** Python does not follow strictly pass by value or pass by reference.

Variables refer to objects. Function behaves whether the object is mutable or immutable
    - Mutable object: can be modified inside the functions. ex: List
    - Immutable object: Remain unchanged. ex: Integer, Strings

```python
def myFun(x):
    x[0] = 20

b = [10, 11, 12, 13]
myFun(b) #modify the 0th index
print(b)

def myFun2(x):
    x = 20

a = 10
myFun2(a) # assign new value to x, but can't modify original value of a.
print(a)
```

### Global and Local Variable
- Global Variable: can be accessed anywhere in the program, include inside function
- Local Variable: define inside the function and exists only during its execution. Can't be accessed from outside the function 

if a variable not found inside a function (local scope), python automatically looks for it in global scope. However, if local and global variable have same name, then local will shadow global variable inside that function.

```python
msg = "Hello from outside the function" #global variable

def greet():
    msg = "Hello from inside the function" #Local variable
    print(msg)

greet()
print(msg)
```

### Recursion in python
- Recursion is programming technique where a function calls itself either directly or indirectly.

use:
1. Breaking Problem into smaller subproblems
2. Mathematical calculation like fibonacci and factorial
3. Tree and Graph traversal and Divide-and-conquer algorithm

- Syntax:
def recursiveFunction(parameters):
    if baseCondition:
        return baseResult
    else:
        return recursiveFunction(modifiedParameter)

Base condition: Stopping condition that prevent infinite loop
Recursive Case: part of the function where it call itself

-Avoid Recursion
    - When problem can be solved easily using loop
    - when recursion depth is large enough to risk stack overflow
    - when performance is critical and function calll overhead matters

- Uses more memory than iteration (loops) due to function call.
- Usually slower

- Types of recursion:
    - Tail Recursion: Recursive call is the last thing function does, so nothing happens after it returns. Can optimize to work like a loop.
    - Non-Tail Recursion: Function does more work after the recursive call returns, so can't optimized into a loop.

```python
def factorial(n):
    if n == 0: #base condition
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```

```python
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(10))
```

```python
def tailFac(n, acc=1):
    if n==0:
        return acc
    else:
        return tailFac(n-1, acc*n)

def nontailFac(n):
    if n==0:
        return 1
    else:
        return n*nontailFac(n-1)
    
print(tailFac(5)) # multiply acc by n before a call, so no extra work is left after recursion
print(nontailFac(5)) # multiplication happen after the recursive call returns, so more work remains after recursion
```

### Lambda Function
- Small anonymous function, meaning they do not have defined name. These are small short-lived function use to pass simple logic to another function
- contain only one expression
- result is returned automatically, no return keyword

- Syntax:
functionName = LambdaKeyword Argument: expression
a = lambda x: x**2

use case:
- condition check
- List comprehension
- Returning Multiple results
- filter()
- map()
- reduce()

```python
a = "Kislay Anand"
upper = lambda x: x.upper() # lambda function takes argument x and return x.upper()
print(upper(a)) #upper(a) applies the lambda to a.
```

```python
check = lambda x: "positive" if x>0 else "negative" if x<0 else "zero"
print(check(5))
print(check(-1))
print(check(0))
```

```python
# List comprehension
func = [lambda arg = x: arg*10 for x in range(1, 5)]
for i in func:
    print(i())
```

```python
# return multiple result
calc = lambda x, y: (x+y, x-y)
res = calc(3, 4)
print(res)
```

```python
c = [1, 2, 3, 4, 5, 6]

#filter
even = filter(lambda x: x%2 == 0, c)
print(list(even))

#map
double = map(lambda x: x*2, c)
print(list(double))

#reduce
from functools import reduce
mul = reduce(lambda x, y: x*y, c)
print(mul)
```

### map(), reduce() and filter()

- syntax:
- map(function, iterable,..) 
    funtion to apply to every element of iterable
    iterable can be one or more iterable objects (list, tuple etc)

- reduce(function, iterable[, initializer])
    function takes two arguments and return single final value
    iterable is the sequence to be reduced
    initializer(optional) is starting value that is placed before first element

    accumulate() : returns an iterator of intermediate results, while reduce() returns only final value.

- filter(function, iterable)
    function: test each element and if return true, keep the element else discard.

```python
s = ['1', '2', '3']
res = map(int, s) # map() applies int() to each element in 's' --> change data type from str to int
print(list(res))
```

```python
def double(val):
    return val*2

a = [1, 2, 3]
res = map(double, a)
print(list(res))
```

```python
# multiple iterable
a = [1, 2, 3]
b = [4, 5, 6]
res = map(lambda x, y: x+y, a, b)
print(list(res))
```

```python
from functools import reduce
from itertools import accumulate

a = [2, 4, 6, 8]
red = reduce(lambda x, y: x + y, a)
print(red)

acc = accumulate(a, lambda x, y: x+y)
print(list(acc))
```

```python
def starts_a(word):
    return word.startswith("a")

li = ["apple", "banana", "avacado", "cherry", "apricot"]
res = filter(starts_a, li)
print(list(res))
```
