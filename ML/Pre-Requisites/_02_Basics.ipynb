### Conditional Statements
if (condition1):
    #if condition1 true, execute this block
elif (condition2):
    #if condition2 true, execute this block
else:
    #if both condition false, execute this

```python
age = 22

if (age<=12):
    print("child")
elif (age<=19):
    print("teenager")
elif (age<=35):
    print("Young adult")
else:
    print("Adult")
```

```python
#Nested conditional statement

age = 70
isMember = True

if age >= 60:
    if isMember:
        print("30% senior discount!")
    else:
        print ("20% senior descount")
else:
    print("Not eligible")
```

```python
#Ternary Conditional Statement --> short way to write if-else, used when choosing beteen two values.
age = 20
s = "Adult" if age>=18 else "Minor" # "statement 1 execute if condition true, else statement 2 execute "
print(s)
```

```python
#Match case statement --> compare value against multiple pattern and execute the matching block of code.
num = 2

match num:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("other number") #--> default case
```

### Loops

- for loop, used to iterate over a sequence such as list, tuple, string or range.
- while loop, repeatedly executes a block as long as given condition remains true. 

>> Loop Control Statement
- Break : used to exit or break out of a loop
- Continue : skip rest of code and force to execute the next iteration
- Pass : when statement is syntactically required but we don't want to execute any code.

```python
num = list(range(0, 10, 2)) #range(x, y, z) x: from, y: till y but y excluded, z: interval. default z = 1 
print(num)
```

```python
n = 4
for i in range(0, n):
    print(i)
```

```python
a = ["one", "two", "three"]
for index in range(len(a)):
    print(a[index])
```

```python
count = 0
while (count < 3):
    print(count)
    count += 1
```

```python
#Nested loop: loop inside loop

for i in range(1, 5):
    for j in range(i):
        print(i, end = " ")
    print()
```

```python
# break
for i in range(5):
    if i == 3:
        break
    print (i)
```

```python
# continue
for i in range(5):
    if i == 3:
        continue
    print (i)
```

```python
# pass
for i in range(5):
    if i == 3:
        pass #placeholder for future code
    print (i)
```

#### Eliminate Loop from python code:


```python
# List Comprehensions : another way to generate lists without the need for explicit loop structure

# without list comprehension
a = []
for i in range(8):
    a.append(i**2)

# with list comprehension
b = [i**2 for i in range(8)]

print("Without:", a)
print("With:", b)
```

```python
# itertools : collection of tools for handling iterators
import itertools

# Flattening a list of lists using loop
a = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
b = []
for sublist in a:
    for i in sublist:
        b.append(i)
print(b)

# Flattening a list of lists using itertools
b = list(itertools.chain(*a))
print(b)
```

```python
# using Numpy

#with loop
num = list(range(1, 6)) 
square = []
for number in num:
    square.append(number**2)
print(square)

#with numpy
import numpy as np
num = np.arange(1, 6)
square = num ** 2
print(square)
```

```python
# generator expression
sq = (n**2 for n in range(10))
for square in sq:
    print(square, end= " ")
```

Using Built in function
- map() : applies a function to all item in an input array
- reduce() : applies a function of two arguments cumulatively to the elements of an iterable
- filter() : creates a list of elements for which the function returns true

```python
numbers = list(range(1, 6))

#map()
squares = list(map(lambda x: x**2, numbers))
print("map: ", squares)

#reduce()
from functools import reduce
product = reduce(lambda x, y: x*y, numbers)
print("reduce:", product)

#filter()
evenNum = list(filter(lambda x: x%2 == 0, numbers))
print("filter: ",evenNum)
```
