```python
final = pd.DataFrame()

for j in range(1, 11):

    url = 'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={}'.format(j)

    headers = {'User-Agent': 'Mozilla/5.0 (Window NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url, headers=headers).text

    company = soup.find_all('div', class_='flex cursor-pointer px-5 pt-5 gap-2 max-md:px-3 max-md:pt-4')
    soup = BeautifulSoup(webpage, 'lxml')

    name = []
    rating = []

    for i in company:
        name.append(i.find('h2').text.strip())
        rating.append(i.find('span', class_='text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]').text.strip())

    d = {'name':name, 'rating':rating}
    df = pd.DataFrame(d)

    final = pd.concat([final, df], ignore_index=True)

```
