# Univariate Analysis:

- analysis of single variable {a data having multiple row, and analysis on one row at a time then know as univariate analysis}
- analysis of two variable {a data having multiple row, and analysis on two row at a time then know as bivariate analysis}
- analysis of multiple variable {a data having multiple row, and analysis on multiple row at a time then know as multivariate analysis}

Data is of two type: -> depending on that we perform EDA
- Numeric
- categorical

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('data/Titanic-Dataset.csv')
```

```python
df.head()
```

## 1. Categorical Data

a. Countplot

```python
sns.countplot(df['Survived'])
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
df['Pclass'].value_counts().plot(kind='bar')
```

b. pie chart

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%.2f')
```

```python
df['Embarked'].value_counts().plot(kind='pie', autopct='%.2f')
```

## 2. Numerical Data

```python
import matplotlib.pyplot as plt
```

a. Histogram -> enable use to use numerical data distribution like categorical data

```python
plt.hist(df['Age'], bins=20, density=True)
```

b. Displot

```python
sns.distplot(df['Age'])
```

c. Boxplot

<outlier>||(Q1-1.5IQR)--------|Q1(25%)<--____IQR_____-->Q2(75%)|---------(Q2+1.5IQR)||<outlier>

```python
sns.boxplot(df['Fare'])
```

```python
plt.boxplot(df['Fare'], vert=False)
```
