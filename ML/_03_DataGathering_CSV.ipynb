1. Optimize Data Types with engine='pyarrow' (Fastest Fix)

By default, pandas uses a C-based parser. If you switch to the PyArrow engine, it can process files up to 5x to 10x faster by using multiple CPU cores.
-  Note: You may need to run pip install pyarrow first if it isn't installed.

df = pd.read_csv("large_file.csv", engine="pyarrow")


2. Only Load the Columns You Need

If your CSV has 50 columns but you only need 3, don't waste memory loading the rest. Use usecols to load specific columns.
python
#Saves massive amounts of memory and time
df = pd.read_csv("large_file.csv", usecols=["User_ID", "Amount", "Date"])


3. Set dtype Manually

Pandas automatically reads through your columns to guess if they are integers, floats, or text. You can skip this heavy calculation by telling pandas the data types upfront using dtype:

#Stops pandas from guessing, making it load instantly
df = pd.read_csv("large_file.csv", dtype={"User_ID": "int32", "Status": "category"})

(Tip: Using "category" instead of text/strings for repeating values saves a massive amount of RAM).

4. Process the File in "Chunks"

If the file is so large that your computer freezes or crashes, use chunksize. This loads the file in bite-sized pieces instead of trying to open everything at once.

- Loads 10,000 rows at a time

chunk_size = 10000
for chunk in pd.read_csv("large_file.csv", chunksize=chunk_size): 
    print(chunk.shape)

```python
#Header Parameter
pd.read_csv('./data/Zomato Restaurant Dataset.csv', header=1) # ->  manipulate header, 1st row become header
```

```python
#use_cols parameter -> get only those columns that we need
pd.read_csv('./data/Zomato Restaurant Dataset.csv', usecols=['Restaurant Name', 'Country Code', 'City'])
```

```python
#squeeze parameter -> need only one column, instead of dataframe, it becomes series
pd.read_csv("./data/Zomato Restaurant Dataset.csv", usecols=['City']).squeeze("columns")
```

```python
#Skip rows/ nrows
pd.read_csv("./data/Zomato Restaurant Dataset.csv", skiprows=[2, 5]) #2nd and 5th has been eliminated
pd.read_csv("./data/Zomato Restaurant Dataset.csv", nrows=20) #gets only 20 rows at a time
```

```python
#Encoding parameter -> if encoding of file is not UTF-8 or something needed
pd.read_csv("./data/Zomato Restaurant Dataset.csv", encoding="latin-1")
```

```python
#skip bad lines
pd.read_csv("./data/Zomato Restaurant Dataset.csv", encoding="latin-1", on_bad_lines="skip") #--> on_bad_line provide 3 way: skip, error; warn
```

```python
# dtype
# pd.read_csv("./data/Zomato Restaurant Dataset.csv").info() #-> gets information
pd.read_csv("./data/Zomato Restaurant Dataset.csv", dtype={'Restaurant ID': float}).info()

```

<b> Converters </b>

```python
def rename(name):
    if (name == "Kislay"):
        return "Kisu"
    else:
        return name   
```
